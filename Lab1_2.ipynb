```python
from collections import deque

# Capacities of jugs
A = 4
B = 3

# Starting state
start = (0, 0)

# BFS queue
queue = deque([(0, 0)])

# To remember visited states
visited = {(0, 0)}

# To store the path
parent = {}
action = {}

goal = None

while queue:

    current = queue.popleft()
    x, y = current

    # Check if we have 2 litres
    if x == 2 or y == 2:
        goal = current
        break

    # All possible operations
    moves = [
        ((4, y), "Fill Jug A"),
        ((x, 3), "Fill Jug B"),
        ((0, y), "Empty Jug A"),
        ((x, 0), "Empty Jug B"),
    ]

    # Pour A into B
    amount = min(x, 3 - y)
    moves.append(
        ((x - amount, y + amount), "Pour A -> B")
    )

    # Pour B into A
    amount = min(y, 4 - x)
    moves.append(
        ((x + amount, y - amount), "Pour B -> A")
    )

    # Add new states
    for new_state, operation in moves:

        if new_state not in visited:

            visited.add(new_state)
            queue.append(new_state)

            parent[new_state] = current
            action[new_state] = operation


# Find the path
path = []

while goal != start:

    path.append(
        (goal, action[goal])
    )

    goal = parent[goal]

path.reverse()


# Print answer
print("WATER JUG PROBLEM")
print("-------------------------")

print("Step 0: (0, 0) - Start")

for i, (state, operation) in enumerate(path, 1):

    print(
        f"Step {i}: {state} - {operation}"
    )

print("\nMinimum steps =", len(path))
```
