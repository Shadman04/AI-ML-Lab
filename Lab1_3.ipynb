```python
# Checking Routes Between Cities using DFS

import networkx as nx
import matplotlib.pyplot as plt


# ---------------------------------------
# Road network
# ---------------------------------------

roads = {
    "Kolkata": ["Howrah"],
    "Howrah": ["Kolkata", "Durgapur"],
    "Durgapur": ["Howrah", "Asansol"],
    "Asansol": ["Durgapur"],

    "Siliguri": ["Jalpaiguri"],
    "Jalpaiguri": ["Siliguri"],

    "Darjeeling": []
}


# ---------------------------------------
# DFS to check a route
# ---------------------------------------

def check_route(city, destination, visited):

    # If we reached the destination
    if city == destination:
        return True

    visited.add(city)

    for next_city in roads[city]:

        if next_city not in visited:

            if check_route(next_city, destination, visited):
                return True

    return False


# ---------------------------------------
# Find connected components
# ---------------------------------------

def find_groups():

    visited = set()
    groups = []

    for city in roads:

        if city not in visited:

            group = []
            stack = [city]

            while stack:

                current = stack.pop()

                if current not in visited:

                    visited.add(current)
                    group.append(current)

                    for next_city in roads[current]:

                        if next_city not in visited:
                            stack.append(next_city)

            groups.append(group)

    return groups


# ---------------------------------------
# Check whether a route exists
# ---------------------------------------

start = "Kolkata"
destination = "Asansol"

print("=" * 50)
print("       CITY ROUTE CHECK USING DFS")
print("=" * 50)

print(f"\nChecking route from {start} to {destination}...")

if check_route(start, destination, set()):
    print("Route found!")
else:
    print("No route found.")


# ---------------------------------------
# Find connected groups
# ---------------------------------------

groups = find_groups()

print("\n" + "=" * 50)
print("          CONNECTED CITY GROUPS")
print("=" * 50)

for number, group in enumerate(groups, 1):
    print(f"Group {number}: {', '.join(group)}")

print(f"\nTotal groups: {len(groups)}")


# ---------------------------------------
# Draw the road network
# ---------------------------------------

G = nx.Graph()

for city in roads:
    for next_city in roads[city]:
        G.add_edge(city, next_city)

# Add Darjeeling even though it has no road
G.add_node("Darjeeling")


# Position of each city
position = {
    "Kolkata": (0, 2),
    "Howrah": (2, 2),
    "Durgapur": (4, 2),
    "Asansol": (6, 2),

    "Siliguri": (2, 0),
    "Jalpaiguri": (4, 0),

    "Darjeeling": (6, 0)
}


# Different colors for different groups
colors = ["skyblue", "lightgreen", "orange"]

city_colors = {}

for i, group in enumerate(groups):
    for city in group:
        city_colors[city] = colors[i]


# Draw the graph
plt.figure(figsize=(12, 6))

nx.draw(
    G,
    position,
    with_labels=True,
    node_color=[city_colors[city] for city in G.nodes()],
    node_size=2500,
    font_size=10,
    font_weight="bold",
    edge_color="gray",
    width=2
)

plt.title(
    "West Bengal City Road Network",
    fontsize=16,
    fontweight="bold"
)

plt.axis("off")
plt.show()
```
