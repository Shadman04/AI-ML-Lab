```
from collections import deque
import matplotlib.pyplot as plt

# -------------------------------------------------
# Jug capacities
# -------------------------------------------------

CAP_A = 4
CAP_B = 3

# Starting state
start = (0, 0)

# Goal: exactly 2 litres in either jug
goal = 2


# -------------------------------------------------
# Function to draw the two jugs
# -------------------------------------------------

def draw_jugs(state, step, operation):
    a, b = state

    fig, ax = plt.subplots(figsize=(7, 5))

    # Jug A
    ax.bar(
        1, a,
        width=0.6,
        color="skyblue",
        edgecolor="black",
        linewidth=2
    )

    # Jug B
    ax.bar(
        2, b,
        width=0.6,
        color="lightgreen",
        edgecolor="black",
        linewidth=2
    )

    # Amount labels
    ax.text(
        1, a + 0.15,
        f"{a} L",
        ha="center",
        fontsize=12,
        fontweight="bold"
    )

    ax.text(
        2, b + 0.15,
        f"{b} L",
        ha="center",
        fontsize=12,
        fontweight="bold"
    )

    ax.set_xticks([1, 2])
    ax.set_xticklabels([
        "Jug A (4 L)",
        "Jug B (3 L)"
    ])

    ax.set_ylim(0, 4.5)
    ax.set_ylabel("Litres")

    ax.set_title(
        f"Step {step}: {operation}",
        fontsize=13,
        fontweight="bold"
    )

    ax.grid(
        axis="y",
        linestyle="--",
        alpha=0.4
    )

    plt.tight_layout()
    plt.show()


# -------------------------------------------------
# Generate all possible operations
# -------------------------------------------------

def get_next_states(state):
    x, y = state

    states = []

    # 1. Fill Jug A
    states.append(
        ((CAP_A, y), "Fill Jug A")
    )

    # 2. Fill Jug B
    states.append(
        ((x, CAP_B), "Fill Jug B")
    )

    # 3. Empty Jug A
    states.append(
        ((0, y), "Empty Jug A")
    )

    # 4. Empty Jug B
    states.append(
        ((x, 0), "Empty Jug B")
    )

    # 5. Pour Jug A -> Jug B
    amount = min(x, CAP_B - y)

    states.append(
        (
            (x - amount, y + amount),
            "Pour Jug A into Jug B"
        )
    )

    # 6. Pour Jug B -> Jug A
    amount = min(y, CAP_A - x)

    states.append(
        (
            (x + amount, y - amount),
            "Pour Jug B into Jug A"
        )
    )

    return states


# -------------------------------------------------
# Breadth First Search
# -------------------------------------------------

queue = deque([start])

visited = {start}

parent = {}
action = {}

goal_state = None

while queue:

    current = queue.popleft()

    x, y = current

    # Check whether goal is reached
    if x == goal or y == goal:
        goal_state = current
        break

    # Generate possible states
    for new_state, operation in get_next_states(current):

        if new_state not in visited:

            visited.add(new_state)
            queue.append(new_state)

            parent[new_state] = current
            action[new_state] = operation


# -------------------------------------------------
# Create solution path
# -------------------------------------------------

path = []

if goal_state is not None:

    state = goal_state

    while state != start:

        path.append(
            (state, action[state])
        )

        state = parent[state]

    path.reverse()


# -------------------------------------------------
# Print solution
# -------------------------------------------------

print("WATER JUG PROBLEM")
print("=" * 40)

print("Jug A capacity = 4 litres")
print("Jug B capacity = 3 litres")
print("Goal = Exactly 2 litres")
print()

if goal_state is None:

    print("No solution found.")

else:

    print("Minimum number of operations:", len(path))
    print()

    print(
        "Step 0: (0, 0) - Both jugs are empty"
    )

    for i, (state, operation) in enumerate(path, 1):

        print(
            f"Step {i}: {state} - {operation}"
        )


    # -------------------------------------------------
    # Display pictures
    # -------------------------------------------------

    print("\nPICTURES OF EACH STEP")
    print("=" * 40)

    # Starting state
    draw_jugs(
        start,
        0,
        "Both jugs are empty"
    )

    # Solution states
    for i, (state, operation) in enumerate(path, 1):

        draw_jugs(
            state,
            i,
            operation
        )
```
